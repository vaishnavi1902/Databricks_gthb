```python
df_marvel = spark.sql("select * from workspace.default.movies where studio = 'Marvel Studios'")
display(df_marvel)
df_marvel.show()
df_otherthanmarvel = spark.sql("select * from workspace.default.movies where studio != 'Marvel Studios'")
display(df_otherthanmarvel)
df_otherthanmarvel.show()
display(df_marvel.count())
df_otherthanmarvel.count()
df_marvel.printSchema()
df_otherthanmarvel.printSchema()
```
